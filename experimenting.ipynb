```python
import pandas as pd

gpay = pd.read_csv("C:\\Users\\Hasan\\Projects\\razorpay buildathon\\Mint-and-Match\\data\\gpay_history.csv")
bank = pd.read_csv("C:\\Users\\Hasan\\Projects\\razorpay buildathon\\Mint-and-Match\\data\\bank_statement.csv")

gpay_cols = list(gpay.columns)
bank_cols = list(bank.columns)

unique_cols = list(set([*gpay_cols , *bank_cols]))



for cols in unique_cols:
    if cols not in bank_cols:
        bank[cols]= None
        
def excract_transition_ID(description : str):
    if not isinstance(description, str):
        return None
    info = description.split('/')
    LEN = 12
    for data in info:
        if len(data) == LEN and data.isdigit():
            return data
    return None

bank["Transaction ID"] = bank["Description"].apply(excract_transition_ID)
    
bank.head(5)
        
```
